Which department uses AI most? X Axis-Major_Category, Y Axis-Weekly_GenAI_Hours

```python
import seaborn as sns
plt.figure(figsize=(10,5))

sns.barplot(
    x='Major_Category',
    y='Weekly_GenAI_Hours',
    data=dataset
)

plt.xticks(rotation=45)

plt.title("Major Category vs Weekly GenAI Hours")

plt.show()
```

Engineering students use AI more. STEM-means Science,Technology,Engineering,Mathematics

Relationship between two numerical columns.  AI Hours vs GPA

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Pre_Semester_GPA',
    y='Post_Semester_GPA',
    data=dataset
)

plt.title("Pre Semester GPA vs Post Semester GPA")

plt.show()
```

More AI usage may improve GPA.

Heatmap-Correlation analysis 

```python
plt.figure(figsize=(12,8))

numeric_dataset = dataset.select_dtypes(include=np.number)

sns.heatmap(
    numeric_dataset.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

Strong relationship between variables.

```python
burnout_count = dataset['Burnout_Risk_Level'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    burnout_count,
    labels=burnout_count.index,
    autopct='%1.1f%%'
)

plt.title("Burnout Risk Level Percentage")

plt.show()
```

```python
major_count = dataset['Major_Category'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    major_count,
    labels=major_count.index,
    autopct='%1.1f%%'
)

plt.title("Major Category Distribution")

plt.show()
```

How many students belong to each burnout category.

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x='Burnout_Risk_Level',
    data=dataset
)

plt.title("Burnout Risk Level")

plt.show()
```

Histogram plot

```python
plt.figure(figsize=(7,5))

sns.histplot(
    dataset['Anxiety_Level_During_Exams'],
    kde=True
)

plt.title("Anxiety Level Distribution")

plt.show()
```

The blue smooth line is called:KDE, Histogram Plot, Used for student mental health and behavioral analysis.
