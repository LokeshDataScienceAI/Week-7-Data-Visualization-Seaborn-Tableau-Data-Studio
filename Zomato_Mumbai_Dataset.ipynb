**Q10) Find the list of Affordable Restaurants?**

**The criteria for Affordable Restaurants would be:**

1) Low Price 2) High Rated

First step will be to find the restaurants with average cost 1/4th the average cost of most expensive restaurant in our dataframe.

Let me explain:-The most expensive restaurant has an average meal cost= 6000. We'll try to stay economical and only pick the
restaurants that are 1/4th of 6000

```python
max_price = zomato_dataset['PRICE'].max() 
one_fourth_price = max_price/4 
one_fourth_price
```

```python
 # Finding list of restaurants that have price less than and equal to 1/4th of the max price i.e Finding Cheap Restaurant
 
aff_rest_dataset = zomato_dataset[['NAME', 'PRICE', 'CUSINE_CATEGORY', 'REGION', 'CUSINE TYPE']] 
aff_rest_dataset = aff_rest_dataset[aff_rest_dataset['PRICE'] <= 1250] 
aff_rest_dataset.sort_values(by='PRICE', inplace=True) 
aff_rest_dataset
```

Now, we'll merge the aff_rest_df with highrate_rest_df to obtain the intersection i.e the list of Affordable Restaurants !!

```python
# Finding the highest rated list of restaurants 
 
highrate_rest_dataset = zomato_dataset[['NAME', 'PRICE', 'CUSINE_CATEGORY', 'REGION', 'CUSINE TYPE','RATING']] 
highrate_rest_dataset = highrate_rest_dataset[highrate_rest_dataset['RATING'] >= 4.5] 
highrate_rest_dataset.sort_values(by='PRICE', inplace=True) 
highrate_rest_dataset
```

```python
highrate_aff_dataset = pd.merge(aff_rest_dataset, highrate_rest_dataset, how='inner', on=['NAME', 'REGION']) 
highrate_aff_dataset = highrate_aff_dataset[['NAME', 'PRICE_x', 'CUSINE_CATEGORY_x', 'REGION', 'CUSINE TYPE_x']] 
highrate_aff_dataset.rename(columns={'NAME':'NAME', 'PRICE_x':'PRICE', 'CUSINE_CATEGORY_x':'CUSINE_CATEGORY',  
                                'REGION':'REGION', 'CUSINE TYPE_x':'CUSINE TYPE'},inplace=True)
```

```python
 # Affordable Restaurants with low price and high rating  
 
highrate_aff_dataset
```

**Q10) Find the list of most Reliable Restaurants?**

**The criteria for most Reliable Restaurants would be:**

1) Low Price 2) High Rated 3) Large No. of Votes

First step will be to find the restaurants with Votes greater than Mean of Votes

```python
mean_votes = zomato_dataset['VOTES'].mean() 
mean_votes
```

```python
# Finding list of restaurants that have Votes greater than and equal to  Mean of Vote 
 
mean_rest_dataset = zomato_dataset[['NAME', 'PRICE', 'CUSINE_CATEGORY', 'REGION', 'CUSINE TYPE', 'VOTES']] 
mean_rest_dataset = mean_rest_dataset[mean_rest_dataset['VOTES'] > 177] 
mean_rest_dataset.sort_values(by='VOTES', inplace=True) 
mean_rest_dataset
```

**These are the most reliable, highest rated and affordable restaurants**

We obtain this dataframe by simply taking the intersection of highrate_aff_df & mean_rest_df

This dataframe obtained below shows the restaurants whose:

_Cost is below_ **1250**,
_Rating is above_ **4.5**,
_Votes are above_ **177**,

```python
reliable_rest_dataset = pd.merge(mean_rest_dataset, highrate_aff_dataset, how='inner', on=['NAME', 'REGION']) 
reliable_rest_dataset = reliable_rest_dataset[['NAME', 'PRICE_x', 'CUSINE_CATEGORY_x', 'REGION', 'CUSINE TYPE_x']] 
reliable_rest_dataset.rename(columns={'NAME':'NAME', 'PRICE_x':'PRICE', 'CUSINE_CATEGORY_x':'CUSINE_CATEGORY',  
                                'REGION':'REGION', 'CUSINE TYPE_x':'CUSINE TYPE'},inplace=True)
```
